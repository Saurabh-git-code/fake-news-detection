```python
import pandas as pd

df = pd.read_csv("cleaned_data.csv")
```

```python
import re

def clean_text(text):
  text = str(text)
  text = text.lower()
  text = re.sub(r'http\S+', '', text)
  text = re.sub(r'[^a-zA-Z]', ' ', text)
  text = re.sub(r'\s+', ' ', text).strip()
  return text

df['clean_text'] = df['text'].apply(clean_text)
```
