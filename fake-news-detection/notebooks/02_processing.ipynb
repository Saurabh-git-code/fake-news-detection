```python
import pandas as pd

df = pd.read_csv("../data/cleaned_data.csv")

df_sample = df.sample(5000, random_state=42).reset_index(drop=True)
```

```python
df.columns
```

```python
import re

def clean_text(text):
  text = str(text)
  text = text.lower()
  text = re.sub(r'http\S+', '', text)
  text = re.sub(r'[^a-zA-Z]', ' ', text)
  text = re.sub(r'\s+', ' ', text).strip()
  return text

df_sample['clean_text'] = df_sample['text'].apply(clean_text)
```

```python
import spacy
nlp = spacy.load("en_core_web_sm",disable=["parser", "ner", "tagger"])
```

```python
docs = list(nlp.pipe(df_sample['clean_text']))
```

```python
df_sample['tokens'] = [
  [token.lemma_ for token in doc if not token.is_stop and not token.is_punct]
  for doc in docs
]
```

```python
df_sample['final_text'] = df_sample['tokens'].apply(lambda x: " ".join(x))
```
